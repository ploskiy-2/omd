**Что нужно сделать:**

На входе информация о конкретном лауреате представляет из себя кучу вложенных друг в друга словарей. Из этой структуры нам надо получить плоский словарь вида {atribute: value}, где atribute --- это один из небольшого множества интересных нам атрибутов. При этом в исходном файле 
1. пути к атрибутам могут иметь разную длину
2. атрибуты могут отсутствовать
3. атрибуты могут нуждаться в дополнительной обработке (например, поле id в исходных словарях записано как строка.)

Поскольку наивная реализация такого обработчика оказывается довольно громоздкой, вам нужно создать файл json_dict_processing.py, в котором реализован высокоуровневый функционал для решения задачи "обработать вложенный json-словарь"

Главное требование к интерфейсу: 
по (вложенному) словарю о лауреате и конфигу (см ниже) надо получить плоский словарь вида {atribute: value}.

Достичь этого можно, например, таким набором функций

* `extract_nested_value(obj, keys)` - извлекает значение из вложенных структур данных по цепочке ключей

* `process_dictionary_with_config(dictionary, config)` - обрабатывает один словарь согласно словару-конфигу (формат конфига см ниже.)

* `process_list_of_dicts_with_config(list_of_dicts, config)` - обрабатывает список словарей, эта функция нужна чтобы обработать список призов.

* `create_processor(config, list_processor=False)` - создает готовый процессор с предзагруженной конфигурацией. В случае, если list_processor == True, внутрь процессора подается process_list_of_dicts_with_config, иначе process_dictionary_with_config.

**Замечание насчет последней функции.**

Планируемое применение этой функции примерно такое:
`laureate_processor = create_processor(PERSON_CONFIG, list_processor=False)`

То есть, create_processor возвращает функцию, которая реализует функционал process_dictionary_with_config но с уже заполненным конфигом. Это повышает читаемость кода и снижает риск перепутать конфиги из-за дублирования кода.

````python
import json

with open('laureates.json', encoding='utf-8') as file:
    laureates = json.load(file)

print(len(laureates['laureates']))
````

````python
from laureates_configs import process_orgs, process_persons


def get_laureate_data(laureate):
    """Определяет тип лауреата и соответствующим образом обрабатывает данные

    Параметры:
        laureate -- сырой словарь лауреата из json

    На выходе:
        Плоский словарь с добавленным полем 'type_' ('person' или 'org')
    """
    if 'knownName' in laureate:
        output = process_persons(laureate)
        output['type_'] = 'person'
    elif 'orgName' in laureate:
        output = process_orgs(laureate)
        output['type_'] = 'org'

    return output


laureates_data = []

for laureate in laureates['laureates']:
    laureates_data.append(get_laureate_data(laureate))
````

````python
from json_dict_processing import process_dictionary_with_config

assert process_dictionary_with_config(
    {'id': '1', 'knownName': {'en': 'Marie Curie'}},
    {'id': (['id'], int), 'name': ['knownName', 'en']},
) == {'id': 1, 'name': 'Marie Curie'}

assert len(laureates_data) == laureates['meta']['count']

assert all(row['type_'] in ('person', 'org') for row in laureates_data)

persons = [row for row in laureates_data if row['type_'] == 'person']
assert all(
    prize['award_year'] > person['birth_year']
    for person in persons
    for prize in person['prizes_relevant']
)

print('Все проверки пройдены')
````

В основном EDA пользуйтесь аккуратным кодом, разбитым на функции и файлы.

````python
from eda import (count_missing, find_duplicates, find_gaps, get_values,
                 missing_share)
from laureates_configs import CONFIG_ORG, CONFIG_PERSON
from prizes_configs import CONFIG_PRIZE

persons = [row for row in laureates_data if row['type_'] == 'person']
orgs = [row for row in laureates_data if row['type_'] == 'org']

print('Всего записей:', len(laureates_data))
print('Из них людей:', len(persons))
print('Из них организаций:', len(orgs))
````

**Вывод:**

Суммарное количество записей людей и организаций в входных данных соответствует общему количеству записей, то есть ни одна запись "не потерялась"

````python
persons_missing = count_missing(persons, list(CONFIG_PERSON))
orgs_missing = count_missing(orgs, list(CONFIG_ORG))
persons_share = missing_share(persons_missing, len(persons))
orgs_share = missing_share(orgs_missing, len(orgs))

total_missing = sum(persons_missing.values()) + sum(orgs_missing.values())
print('Все поля заполнены:', total_missing == 0)
print()

print('Пропуски у людей (поле, количество, доля):')
for field in persons_missing:
    if persons_missing[field] > 0:
        print(f'{field}, {persons_missing[field]}, {persons_share[field]}')
print()

print('Пропуски у организаций (поле, количество, доля):')
for field in orgs_missing:
    if orgs_missing[field] > 0:
        print(f'{field}, {orgs_missing[field]}, {orgs_share[field]}')
````

**Вывод:**

Пропуски в данных присутствуют:

1. По 3 незаполненные записи для полей country_birth и country_now. При этом пропуски составляют менее 1% от общего числа записей, поэтому здесь нет ничего страшного

2. У организаций пропуски в полях founded_year, country_founded и country_now. Год основания не указан в единственном случае - не критическая проблема. Поля country_founded и country_now не указаны в ~14% случае, но всего организаций 28, поэтому то, что 4 без страны это нормально. Можем подробнее посмотреть, что это за организации:

````python
print('Организации без страны основания (название, год основания):')
for row in orgs:
    if row['country_founded'] is None:
        print(row['name'], row['founded_year'])
````

````python
all_prizes = []
for row in laureates_data:
    all_prizes.extend(row['prizes_relevant'])

prizes_missing = count_missing(all_prizes, list(CONFIG_PRIZE))
prizes_share = missing_share(prizes_missing, len(all_prizes))

print('Всего призов:', len(all_prizes))
print()

if sum(prizes_missing.values()) == 0:
    print('Пропусков в призах нет')
else:
    print('Пропуски в призах (поле, количество, доля):')
    for field in prizes_missing:
        if prizes_missing[field] > 0:
            print(field, prizes_missing[field], prizes_share[field])
````

**Вывод:**

Призы заполнены без пропусков

````python
ids = get_values(laureates_data, 'id')
laureate_by_id = {row['id']: row for row in laureates_data}

id_extremes = [('Минимальный', min(ids)), ('Максимальный', max(ids))]
for label, laureate_id in id_extremes:
    row = laureate_by_id[laureate_id]
    award_years = get_values(row['prizes_relevant'], 'award_year')
    print(label, 'id:', laureate_id)
    print('    лауреат:', row['name'])
    print('    годы награждения:', award_years)
````

````python
duplicate_ids = find_duplicates(ids)
id_gaps = find_gaps(ids)
missing_ids_count = sum(right - left - 1 for left, right in id_gaps)

print('Повторяющиеся id:', sorted(duplicate_ids))
print('Пар соседних id с разницей больше 1:', len(id_gaps))
print('Всего пропущено номеров:', missing_ids_count)
print('Лауреатов по данным meta:', laureates['meta']['count'])
print('Пары:', id_gaps)
````

- Выделите самые частые топ-50 слов. Какие из этих слов самые частые в принципе в английском языке (такие, как for, the и другие), а какие кажутся важными именно в контексте вручения Нобелевской премии (например, 'chemistry')?
