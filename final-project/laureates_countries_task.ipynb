**Что нужно сделать:**

На входе информация о конкретном лауреате представляет из себя кучу вложенных друг в друга словарей. Из этой структуры нам надо получить плоский словарь вида {atribute: value}, где atribute --- это один из небольшого множества интересных нам атрибутов. При этом в исходном файле 
1. пути к атрибутам могут иметь разную длину
2. атрибуты могут отсутствовать
3. атрибуты могут нуждаться в дополнительной обработке (например, поле id в исходных словарях записано как строка.)

Поскольку наивная реализация такого обработчика оказывается довольно громоздкой, вам нужно создать файл json_dict_processing.py, в котором реализован высокоуровневый функционал для решения задачи "обработать вложенный json-словарь"

Главное требование к интерфейсу: 
по (вложенному) словарю о лауреате и конфигу (см ниже) надо получить плоский словарь вида {atribute: value}.

Достичь этого можно, например, таким набором функций

* `extract_nested_value(obj, keys)` - извлекает значение из вложенных структур данных по цепочке ключей

* `process_dictionary_with_config(dictionary, config)` - обрабатывает один словарь согласно словару-конфигу (формат конфига см ниже.)

* `process_list_of_dicts_with_config(list_of_dicts, config)` - обрабатывает список словарей, эта функция нужна чтобы обработать список призов.

* `create_processor(config, list_processor=False)` - создает готовый процессор с предзагруженной конфигурацией. В случае, если list_processor == True, внутрь процессора подается process_list_of_dicts_with_config, иначе process_dictionary_with_config.

**Замечание насчет последней функции.**

Планируемое применение этой функции примерно такое:
`laureate_processor = create_processor(PERSON_CONFIG, list_processor=False)`

То есть, create_processor возвращает функцию, которая реализует функционал process_dictionary_with_config но с уже заполненным конфигом. Это повышает читаемость кода и снижает риск перепутать конфиги из-за дублирования кода.


Помимо этого нужно создать файлы-конфиги для данных по призам, лауреатам-людям, и лауреатам-организациям.

В случае, если атрибут можно взять в итоговый датасет сырым, в values() словаря-конфига будет список path_list.

````python
import json

with open('laureates.json', encoding='utf-8') as file:
    laureates = json.load(file)

print(len(laureates['laureates']))
````

````python
from laureates_configs import process_orgs, process_persons


def get_laureate_data(laureate):
    """Определяет тип лауреата и соответствующим образом обрабатывает данные

    Параметры:
        laureate -- сырой словарь лауреата из json

    На выходе:
        Плоский словарь с добавленным полем 'type_' ('person' или 'org')
    """
    if 'knownName' in laureate:
        output = process_persons(laureate)
        output['type_'] = 'person'
    elif 'orgName' in laureate:
        output = process_orgs(laureate)
        output['type_'] = 'org'

    return output


laureates_data = []

for laureate in laureates['laureates']:
    laureates_data.append(get_laureate_data(laureate))
````

````python
from json_dict_processing import process_dictionary_with_config

assert process_dictionary_with_config(
    {'id': '1', 'knownName': {'en': 'Marie Curie'}},
    {'id': (['id'], int), 'name': ['knownName', 'en']},
) == {'id': 1, 'name': 'Marie Curie'}

assert len(laureates_data) == laureates['meta']['count']

assert all(row['type_'] in ('person', 'org') for row in laureates_data)

persons = [row for row in laureates_data if row['type_'] == 'person']
assert all(
    prize['award_year'] > person['birth_year']
    for person in persons
    for prize in person['prizes_relevant']
)

print('Все проверки пройдены')
````

````python
from eda import (count_missing, find_duplicates, find_gaps, get_values,
                 missing_share)
from laureates_configs import CONFIG_ORG, CONFIG_PERSON
from prizes_configs import CONFIG_PRIZE

persons = [row for row in laureates_data if row['type_'] == 'person']
orgs = [row for row in laureates_data if row['type_'] == 'org']

print('Всего записей:', len(laureates_data))
print('Из них людей:', len(persons))
print('Из них организаций:', len(orgs))
````

**Вывод:**

Суммарное количество записей людей и организаций в входных данных соответствует общему количеству записей, то есть ни одна запись "не потерялась"

````python
persons_missing = count_missing(persons, list(CONFIG_PERSON))
orgs_missing = count_missing(orgs, list(CONFIG_ORG))
persons_share = missing_share(persons_missing, len(persons))
orgs_share = missing_share(orgs_missing, len(orgs))

total_missing = sum(persons_missing.values()) + sum(orgs_missing.values())
print('Все поля заполнены:', total_missing == 0)
print()

print('Пропуски у людей (поле, количество, доля):')
for field in persons_missing:
    if persons_missing[field] > 0:
        print(f'{field}, {persons_missing[field]}, {persons_share[field]}')
print()

print('Пропуски у организаций (поле, количество, доля):')
for field in orgs_missing:
    if orgs_missing[field] > 0:
        print(f'{field}, {orgs_missing[field]}, {orgs_share[field]}')
````

**Вывод:**

Пропуски в данных присутствуют:

1. По 3 незаполненные записи для полей country_birth и country_now. При этом пропуски составляют менее 1% от общего числа записей, поэтому здесь нет ничего страшного

2. У организаций пропуски в полях founded_year, country_founded и country_now. Год основания не указан в единственном случае - не критическая проблема. Поля country_founded и country_now не указаны в ~14% случае, но всего организаций 28, поэтому то, что 4 без страны это нормально. Можем подробнее посмотреть, что это за организации:

````python
print('Организации без страны основания (название, год основания):')
for row in orgs:
    if row['country_founded'] is None:
        print(row['name'], row['founded_year'])
````

````python
all_prizes = []
for row in laureates_data:
    all_prizes.extend(row['prizes_relevant'])

prizes_missing = count_missing(all_prizes, list(CONFIG_PRIZE))
prizes_share = missing_share(prizes_missing, len(all_prizes))

print('Всего призов:', len(all_prizes))
print()

if sum(prizes_missing.values()) == 0:
    print('Пропусков в призах нет')
else:
    print('Пропуски в призах (поле, количество, доля):')
    for field in prizes_missing:
        if prizes_missing[field] > 0:
            print(field, prizes_missing[field], prizes_share[field])
````

**Вывод:**

Призы заполнены без пропусков

````python
ids = get_values(laureates_data, 'id')
laureate_by_id = {row['id']: row for row in laureates_data}

id_extremes = [('Минимальный', min(ids)), ('Максимальный', max(ids))]
for label, laureate_id in id_extremes:
    row = laureate_by_id[laureate_id]
    award_years = get_values(row['prizes_relevant'], 'award_year')
    print(label, 'id:', laureate_id)
    print('    лауреат:', row['name'])
    print('    годы награждения:', award_years)
````

### Общие атрибуты для анализа

````python
from aggregations import (count_values, max_value, mean_value, median_value,
                          min_value, top_n)
from atribute_manager import add_attribute
from dataset_helpers import (apply_aggregation, filter_records,
                             group_by_attributes)
from plots import plot_bar, plot_line


def get_first_award_year(row):
    """Год первой премии - самый ранний award_year среди призов"""
    return min_value(row['prizes_relevant'], 'award_year')


def get_first_category(row):
    """Категория приза, полученного в год первой премии"""
    first_year = get_first_award_year(row)
    for prize in row['prizes_relevant']:
        if prize['award_year'] == first_year:
            return prize['category_en']


def get_decade(row):
    """Десятилетие первой премии"""
    return row['first_award_year'] // 10 * 10


def get_age_at_first_prize(row):
    """Возраст лауреата на момент первой премии"""
    return row['first_award_year'] - row['birth_year']


def is_person(row):
    """Проверяет, что лауреат - человек"""
    return row['type_'] == 'person'
````

````python
analysis_data = add_attribute(
    laureates_data, 'first_award_year', get_first_award_year
)
analysis_data = add_attribute(
    analysis_data, 'first_category', get_first_category
)
analysis_data = add_attribute(analysis_data, 'decade', get_decade)

analysis_persons = filter_records(analysis_data, is_person)
analysis_persons = add_attribute(
    analysis_persons, 'age_at_first_prize', get_age_at_first_prize
)

print('Пример записи с новыми полями:')
example = analysis_persons[0]
for field in ['name', 'first_award_year', 'first_category', 'decade',
              'age_at_first_prize']:
    print(field, '-', example[field])
````

### Тема 1. Возраст лауреатов на момент первой премии

````python
youngest_age = min_value(analysis_persons, 'age_at_first_prize')
oldest_age = max_value(analysis_persons, 'age_at_first_prize')

print('Лауреатов:', count_values(analysis_persons, 'id'))
print('Средний возраст:',
      round(mean_value(analysis_persons, 'age_at_first_prize'), 1))
print('Медианный возраст:',
      median_value(analysis_persons, 'age_at_first_prize'))
print()

for row in analysis_persons:
    if row['age_at_first_prize'] == youngest_age:
        print('Самый молодой:', row['name'], '-', youngest_age, 'лет,',
              row['first_award_year'], 'год,', row['first_category'])
    if row['age_at_first_prize'] == oldest_age:
        print('Самый пожилой:', row['name'], '-', oldest_age, 'лет,',
              row['first_award_year'], 'год,', row['first_category'])
````

**Вывод:**
Средний возраст и медианный возраст практически схожи - это означает, что "очень молодые" или "очень старые" лауреаты не искажают среднее. То есть обычно награду получают в "зрелом" возрасте около 60 лет

````python
groups_by_category = group_by_attributes(analysis_persons, ['first_category'])
count_by_category = apply_aggregation(groups_by_category, count_values, 'id')
mean_age_by_category = apply_aggregation(
    groups_by_category, mean_value, 'age_at_first_prize'
)
median_age_by_category = apply_aggregation(
    groups_by_category, median_value, 'age_at_first_prize'
)

print('Категория              | Лауреатов |',
      'Средний возраст | Медианный возраст')
print('-' * 75)
for key, mean_age in top_n(mean_age_by_category, len(mean_age_by_category)):
    count = count_by_category[key]
    median_age = median_age_by_category[key]
    print(f'{key[0]:<23}| {count:>9} | '
          f'{mean_age:>15.1f} | {median_age:>17.1f}')

plot_bar(mean_age_by_category,
         'Средний возраст, при первой премии по категориям', 'Возраст, лет')
````

**Вывод:**

Старше всех лауреаты по экономике и литературе. Моложе всех физики, затем медицина и химия. Разница между крайними категориями - почти 10 лет

Возможно это связано с тем, что в естественных науках премию часто дают за конкретное открытие, а в экономике и литературе - скорее за вклад в науку

В литературе медиана выше среднего значения, значит есть несколько "молодых писателей", которые тянут среднее вниз 
В физике наоборот - медиана 56 ниже среднего 57.8, то есть среднее поднимают несколько "старых ученых"

````python
groups_by_decade = group_by_attributes(analysis_persons, ['decade'])
mean_age_by_decade = apply_aggregation(
    groups_by_decade, mean_value, 'age_at_first_prize'
)
median_age_by_decade = apply_aggregation(
    groups_by_decade, median_value, 'age_at_first_prize'
)

print('Десятилетие | Средний возраст | Медианный возраст')
print('-' * 49)
for key in sorted(mean_age_by_decade):
    mean_age = mean_age_by_decade[key]
    median_age = median_age_by_decade[key]
    print(f'{key[0]:<12}| {mean_age:>15.1f} | {median_age:>17.1f}')

plot_line(mean_age_by_decade,
          'Средний возраст при первой премии по десятилетиям', 'Возраст, лет')
````

**Вывод:**
Начиная с 1950х годов возраст лауреатов почти непрерывно растет: 59.3 в 1970х, 62.1 в 1990х, 65.9 в 2000х и 69.4 в 2020х. За 90 лет он вырос почти на 18 лет

Возможные причины:
- между открытием и его признанием проходит все больше времени
- путь к крупному научному результату стал дольше

### Тема 2. Женщины-лауреаты

````python
def is_female(row):
    """Проверяет, что лауреат - женщина"""
    return row['gender'] == 'female'


women = filter_records(analysis_persons, is_female)
women_count = count_values(women, 'id')
persons_count = count_values(analysis_persons, 'id')

print('Женщин-лауреатов:', women_count, 'из', persons_count)
print('Доля женщин:', round(women_count / persons_count, 3))
````

**Вывод:**

Среди 1000 лауреатов женщин всего 69, то есть 6.9%. То есть чаще Нобелевскую премию получают мужчины

````python
women_groups_by_decade = group_by_attributes(women, ['decade'])
women_by_decade = apply_aggregation(women_groups_by_decade, count_values, 'id')
persons_by_decade = apply_aggregation(groups_by_decade, count_values, 'id')

# в некоторых десятилетиях женщин нет, для них ставим 0
women_count_by_decade = {}
women_share_by_decade = {}
for key in persons_by_decade:
    women_count_by_decade[key] = women_by_decade.get(key, 0)
    women_share_by_decade[key] = (
        women_count_by_decade[key] / persons_by_decade[key]
    )

print('Десятилетие | Всего лауреатов | Женщин | Доля женщин')
print('-' * 52)
for key in sorted(persons_by_decade):
    total = persons_by_decade[key]
    women_count = women_count_by_decade[key]
    share = women_share_by_decade[key]
    print(f'{key[0]:<12}| {total:>15} | {women_count:>6} | {share:>11.3f}')

plot_line(women_count_by_decade,
          'Количество женщин-лауреатов по десятилетиям', 'Количество')
plot_line(women_share_by_decade,
          'Доля женщин среди лауреатов по десятилетиям', 'Доля')
````

**Вывод:**

До 1990х женщины получали премию редко, от 0 до 4 человек за десятилетие, доля составляла от 0 до 7.5%

С 1990х начинается рост наград у женщин. В 2020х уже 16 лауреаток, что составляло 19.8%, то есть почти каждый пятый лауреат

В данном случае, смотреть стоит именно на долю, так как лауреатов вообще со временем стало больше (56 в 1900х против 100–120 в последние десятилетия), поэтому рост количества частично объясняется общим ростом лауреатов. Но доля то же выросла, примерно с 4-5% до 20% --> значит рост реальный

````python
women_groups_by_category = group_by_attributes(women, ['first_category'])
women_by_category = apply_aggregation(
    women_groups_by_category, count_values, 'id'
)

print('Категория              | Женщин за все годы')
print('-' * 42)
for key, count in top_n(women_by_category, len(women_by_category)):
    print(f'{key[0]:<23}| {count:>18}')

plot_bar(women_by_category,
         'Женщины-лауреаты по категориям за все годы', 'Количество')
````

**Вывод:**

Больше всего женщин-лауреатов в премии мира (21) и литературе (19), дальше идет медицина. В естественных и точных науках их меньше

````python
women_by_decade_category = apply_aggregation(
    group_by_attributes(women, ['decade', 'first_category']),
    count_values, 'id'
)

print('Десятилетие | Категория              | Женщин')
print('-' * 45)
for key in sorted(women_by_decade_category):
    count = women_by_decade_category[key]
    print(f'{key[0]:<12}| {key[1]:<23}| {count:>6}')
````

**Вывод:**

Первые полвека женщины получали премии в основном за литературу и премию мира. В естественных науках были единичные случаи:
- Мария Кюри - физика
- Ирен Жолио-Кюри - химия
- Герти Кори - медицина

После 2000х женщины появляются почти во всех категориях в каждом десятилетии:
- Элинор Остром - экономика
- Эстер Дюфло - экономика
- Клаудия Голдин - экономика
- Донна Стрикленд - физика
- Андреа Гез - физика
- Анн Л’Юилье - физика

````python
def is_early(row):
    """Первая премия получена до 1950 года"""
    return row['first_award_year'] < 1950


def is_recent(row):
    """Первая премия получена в 2010 году или позже"""
    return row['first_award_year'] >= 2010


early_women = filter_records(women, is_early)
recent_women = filter_records(women, is_recent)
early_by_category = apply_aggregation(
    group_by_attributes(early_women, ['first_category']), count_values, 'id'
)
recent_by_category = apply_aggregation(
    group_by_attributes(recent_women, ['first_category']), count_values, 'id'
)

print('До 1950 года')
print('Категория              | Женщин')
print('-' * 31)
for key, count in top_n(early_by_category, len(early_by_category)):
    print(f'{key[0]:<23}| {count:>6}')
print()
print('С 2010 года')
print('Категория              | Женщин')
print('-' * 31)
for key, count in top_n(recent_by_category, len(recent_by_category)):
    print(f'{key[0]:<23}| {count:>6}')
````

**Вывод:**

До 1950го года больше всего женщин получили премию по литературе (5 из 11), а с 2010го года лидирует премия мира (9 из 29)

В последние годы женщин стало заметно больше и в естественных науках. До 1950 года на химию, медицину и физику приходилось 3 женщины из 11 (что составляло 27%), а с 2010 года приходится 11 из 29 (что составляет 38%)
